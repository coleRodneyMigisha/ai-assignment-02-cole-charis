```python
import pandas as pd
```

```python
#create the two CSVs from the sheets

def createCSVs(xls):
    for sheet_name in xls.sheet_names:
        df = pd.read_excel(xls, sheet_name)
        df.to_csv(f'{sheet_name}.csv', index=False)
```

```python
xls = pd.ExcelFile('dsca.xlsx')

createCSVs(xls)
```
